###Handling Missing Values

#####Check missing values before cleaning

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(
    missing_summary[
        missing_summary["Missing Values"] > 0
    ].sort_values(
        by="Missing Values",
        ascending=False
    )
)
```

The missing-value analysis identified columns containing unavailable information. Missing categorical values are handled separately while numerical and date fields are investigated before applying any imputation or conversion.

####Handle Categorical Missing Values

```python
categorical_cols = [
    "Invoice Description",
    "Vendor/Payee"
]

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")
```

Missing values in categorical fields were replaced with "Unknown" to preserve the corresponding transaction records while clearly indicating that the original information was unavailable.

####Handling Numerical Missing Values

```python
numeric_missing = df.select_dtypes(include=np.number).isnull().sum()
display(numeric_missing[numeric_missing > 0])
```

No missing values were identified in the numerical columns, so numerical imputation was not required.

```python
df.isnull().sum()
```

The missing-value treatment was successful, and no remaining missing values were identified in the selected fields.

###Handling Duplicate Rows

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print("Duplicate records available for inspection:")
display(duplicate_rows.head(20))
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Dataset Shape after duplicate removal:", df.shape)
print("Duplicate rows after removal:", df.duplicated().sum())
```

The identified duplicate records were inspected and confirmed as exact duplicate rows before removal. The confirmed duplicates were removed to prevent repeated records from affecting subsequent financial analysis.

####Converting Data Types

#####Payment Amount

```python
if df["Payment Amount"].dtype == "object":
    df["Payment Amount"] = (
        df["Payment Amount"]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .str.strip()
    )

df["Payment Amount"] = pd.to_numeric(
    df["Payment Amount"],
    errors="coerce"
)

print("Payment Amount data type:",
      df["Payment Amount"].dtype)

print("Missing Payment Amounts after conversion:",
      df["Payment Amount"].isnull().sum())
```

#####Payment Date

```python
df["Payment Date"] = pd.to_datetime(
    df["Payment Date"],
    dayfirst=True,
    errors="coerce"
)

print("Payment Date data type after conversion:",
      df["Payment Date"].dtype)

print("Missing/invalid dates after conversion:",
      df["Payment Date"].isnull().sum())
```

The Payment Date column was converted from object format to datetime format using Pandas. Day-first date interpretation was specified to correctly interpret the date values, and the conversion was checked for missing or invalid dates. No missing dates were created during the conversion.

###Outlier Detection Using IQR

```python
Q1 = df["Payment Amount"].quantile(0.25)
Q3 = df["Payment Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# identify Outliers
outliers = df[
    (df["Payment Amount"] < lower_limit) |
    (df["Payment Amount"] > upper_limit)
]
print("Number of outliers:", len(outliers))
```

```python
display(
    outliers[
        [
            "Payment Date",
            "Vendor/Payee",
            "City Department Account",
            "Payment Amount"
        ]
    ].head(20)
)
```

The IQR method identified payment records outside the calculated statistical limits. These observations were inspected but were not automatically removed because statistical outliers are not necessarily data errors. Large government payments may represent legitimate financial transactions.

### 1. Descriptive Analysis of Payment Amounts

This analysis examines the central tendency, variability, range, and distribution of payment amounts. It also identifies the number of positive, zero, and negative transactions without modifying the original dataset.

```python
payment_amount = df["Payment_Amount"].dropna()

print("DATASET OVERVIEW")
print("-" * 40)
print("Dataset shape:", df.shape)
print("Number of payment records:", len(payment_amount))

print("\nDESCRIPTIVE STATISTICS")
print("-" * 40)

display(
    payment_amount
    .describe()
    .to_frame(name="Payment Amount")
    .round(2)
)

print("\nNET PAYMENT TOTAL")
print("-" * 40)
print(
    "Net total (including negative values):",
    round(payment_amount.sum(), 2)
)

print("\nTRANSACTIONS BY PAYMENT SIGN")
print("-" * 40)

sign_summary = pd.DataFrame({
    "Transaction Count": [
        (payment_amount < 0).sum(),
        (payment_amount == 0).sum(),
        (payment_amount > 0).sum()
    ]
}, index=["Negative payments", "Zero payments", "Positive payments"])

sign_summary["Percentage"] = (
    sign_summary["Transaction Count"]
    / len(payment_amount)
    * 100
)

display(sign_summary.round(2))

print("\nDISTRIBUTION SHAPE")
print("-" * 40)
print("Skewness:", round(payment_amount.skew(), 2))
```

```python
positive_payments = df.loc[
    df["Payment_Amount"] > 0,
    "Payment_Amount"
].dropna()

print("Positive payments:", len(positive_payments))
print("Negative payments:", (df["Payment_Amount"] < 0).sum())
print("Zero payments:", (df["Payment_Amount"] == 0).sum())

bin_edges = np.logspace(
    np.log10(positive_payments.min()),
    np.log10(positive_payments.max()),
    51
)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.histplot(
    x=positive_payments,
    bins=bin_edges,
    ax=axes[0]
)
axes[0].set_xscale("log")
axes[0].set_title("Distribution of Positive Payment Amounts")
axes[0].set_xlabel("Payment Amount (log scale)")
axes[0].set_ylabel("Number of Transactions")

# Boxplot
sns.boxplot(
    x=positive_payments,
    ax=axes[1],
    showfliers=False
)
axes[1].set_xscale("log")
axes[1].set_title("Spread of Positive Payment Amounts")
axes[1].set_xlabel("Payment Amount (log scale)")

plt.tight_layout()
plt.show()
```

The payment amounts are highly right-skewed, with the mean (6,725.57) exceeding the median (364.64), indicating that a small number of large transactions influence the average. The middle 50% of recorded payments range from 105.00 to 1,699.26. The dataset contains 3,471 negative payments (0.94%) and 12 zero-value payments (less than 0.01%). These values are retained unchanged pending investigation of their meaning.

####Payment Method Composition

```python
# Count transactions by payment method.
# Missing method labels, if any, are reported as "Missing".
method_series = df["payment_method"].fillna("Missing")

method_counts = (
    method_series
    .value_counts()
    .rename_axis("Payment Method")
    .reset_index(name="Transaction Count")
)

# Calculate the percentage of transactions
method_counts["Percentage"] = (
    method_counts["Transaction Count"]
    / method_counts["Transaction Count"].sum()
    * 100
)

print("TRANSACTIONS BY PAYMENT METHOD")
display(method_counts.round(2))

# Doughnut chart
fig, ax = plt.subplots(figsize=(8, 6))

wedges, texts, autotexts = ax.pie(
    method_counts["Transaction Count"],
    labels=method_counts["Payment Method"],
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.75,
    wedgeprops={
        "width": 0.42,
        "edgecolor": "white"
    }
)

ax.set_title("Share of Payment Transactions by Method")
ax.text(
    0, 0,
    f"{method_counts['Transaction Count'].sum():,}\nTransactions",
    ha="center",
    va="center",
    fontsize=12
)

plt.tight_layout()
plt.show()
```

System checks account for the majority of transactions (88.52%), while e-payables represent 11.48%. This indicates that system checks are the predominant payment method by transaction count. However, these percentages reflect transaction frequency, not total payment value.

####Total Payment Amount by Payment Method

```python
payment_totals = (
    df.assign(
        payment_method=df["payment_method"].fillna("Missing")
    )
    .groupby("payment_method")["Payment_Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("TOTAL PAYMENT AMOUNT BY PAYMENT METHOD")
display(payment_totals.to_frame(name="Total Payment Amount").round(2))

payment_totals_millions = payment_totals / 1_000_000

plt.figure(figsize=(9, 5))

bars = plt.bar(
    payment_totals_millions.index,
    payment_totals_millions.values
)

plt.title("Total Payment Amount by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Payment Amount (Millions)")

for bar in bars:
    value = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.1f}M",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

System checks account for approximately 95.99% of the total recorded payment value, amounting to 2.395 billion, while e-payables account for approximately 4.01%, or 100.14 million. This indicates that system checks account for the larger share of recorded payment value.

####Monthly Payment Trends

```python
monthly_data = df[
    ["Payment_Date", "Payment_Amount"]
].dropna(subset=["Payment_Date"]).copy()

# Group payment dates by calendar month
monthly_data["Month"] = (
    monthly_data["Payment_Date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

# Calculate monthly totals and transaction counts
monthly_totals_all = (
    monthly_data
    .groupby("Month")["Payment_Amount"]
    .sum()
    .sort_index()
)

monthly_counts = (
    monthly_data
    .groupby("Month")["Payment_Amount"]
    .count()
)

first_month = monthly_totals_all.index.min()
last_month = monthly_totals_all.index.max()

monthly_totals = monthly_totals_all.loc[
    (monthly_totals_all.index > first_month) &
    (monthly_totals_all.index < last_month)
]

print("MONTHLY PAYMENT TREND")
print("-" * 40)
print("Months with recorded transactions:", len(monthly_totals_all))
print("First recorded month:", first_month.strftime("%B %Y"))
print("Transactions in first month:", monthly_counts.loc[first_month])
print("Last recorded month:", last_month.strftime("%B %Y"))
print("Transactions in last month:", monthly_counts.loc[last_month])

print("\nMONTHS INCLUDED IN THE CHART")
print("Number of observed months:", len(monthly_totals))
print("First month:", monthly_totals.index.min().strftime("%B %Y"))
print("Last month:", monthly_totals.index.max().strftime("%B %Y"))

print("\nHighest monthly total in the chart:")
print(
    monthly_totals.idxmax().strftime("%B %Y"),
    round(monthly_totals.max(), 2)
)

monthly_totals_millions = monthly_totals / 1_000_000

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_totals_millions.index,
    monthly_totals_millions.values,
    marker="o",
    markersize=3,
    linewidth=1.5
)

plt.title("Monthly Payment Trends (Excluding Partial Boundary Months)")
plt.xlabel("Month")
plt.ylabel("Total Payment Amount (Millions)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

Monthly payment totals fluctuate over time, with January 2024 recording the highest total at approximately 37.15 million. July 2014 and September 2026 were excluded from the chart because they contained only one and six transactions, respectively. The chart covers July 2016 through August 2026, as no transactions were recorded in 2015.

####Annual Payment Trends

```python
yearly_data = df[
    ["Payment_Date", "Payment_Amount"]
].dropna(subset=["Payment_Date"]).copy()

yearly_data["Year"] = yearly_data["Payment_Date"].dt.year

# Calculate total payment amount by year
yearly_totals = (
    yearly_data
    .groupby("Year")["Payment_Amount"]
    .sum()
    .sort_index()
)

# Compare complete calendar years only (2017–2025)
annual_totals = yearly_totals.loc[
    (yearly_totals.index >= 2017) &
    (yearly_totals.index <= 2025)
]

print("ANNUAL PAYMENT TOTALS (2017–2025)")

display(
    annual_totals
    .to_frame(name="Total Payment Amount")
    .round(2)
)

annual_totals_millions = annual_totals / 1_000_000

plt.figure(figsize=(12, 6))

bars = plt.bar(
    annual_totals_millions.index.astype(str),
    annual_totals_millions.values
)

plt.title("Annual Payment Trends (2017–2025)")
plt.xlabel("Calendar Year")
plt.ylabel("Total Payment Amount (Millions)")

for bar in bars:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value,
        f"{value:.1f}M",
        ha="center",
        va="bottom",
        fontsize=8
    )

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Annual payment totals generally increased from 2017 to 2025, reaching approximately 305.27 million in 2025. Temporary declines occurred in 2020 and 2024. The analysis excludes 2014, which contains only one transaction, 2015, which has no recorded transactions, and 2016, which represents a partial calendar year.

#### Correlation Between Monthly Transaction Counts and Payment Amounts

```python
# Create a temporary dataset for monthly analysis
correlation_data = df[
    ["Payment_Date", "Payment_Amount"]
].dropna(subset=["Payment_Date"]).copy()

# Group transactions by calendar month
correlation_data["Month"] = (
    correlation_data["Payment_Date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

monthly_correlation = (
    correlation_data
    .groupby("Month")
    .agg(
        Transaction_Count=("Payment_Amount", "count"),
        Total_Payment_Amount=("Payment_Amount", "sum")
    )
    .reset_index()
)

first_month = monthly_correlation["Month"].min()
last_month = monthly_correlation["Month"].max()

monthly_correlation = monthly_correlation[
    (monthly_correlation["Month"] > first_month) &
    (monthly_correlation["Month"] < last_month)
].copy()

# Correlation coefficients
pearson_corr = monthly_correlation[
    "Transaction_Count"
].corr(
    monthly_correlation["Total_Payment_Amount"],
    method="pearson"
)

spearman_corr = monthly_correlation[
    "Transaction_Count"
].corr(
    monthly_correlation["Total_Payment_Amount"],
    method="spearman"
)

print("CORRELATION ANALYSIS")
print("-" * 40)
print("Months analyzed:", len(monthly_correlation))
print("Pearson correlation:", round(pearson_corr, 3))
print("Spearman correlation:", round(spearman_corr, 3))
```

This indicates that months with more transactions tend to have higher payment totals, although the relationship varies. The results show association, not causation, and may be influenced by unusual months such as December 2018.

#### Typical Payment Totals by Calendar Month

```python
seasonal_data = df[
    ["Payment_Date", "Payment_Amount"]
].dropna(subset=["Payment_Date"]).copy()

seasonal_data["Year"] = seasonal_data["Payment_Date"].dt.year
seasonal_data["Month"] = seasonal_data["Payment_Date"].dt.month

# Complete calendar years only
seasonal_data = seasonal_data[
    (seasonal_data["Year"] >= 2017) &
    (seasonal_data["Year"] <= 2025)
].copy()

# Total payments for each year and calendar month
monthly_totals_seasonal = (
    seasonal_data
    .groupby(["Year", "Month"])["Payment_Amount"]
    .sum()
    .reset_index()
)

# The median monthly total for each calendar month
typical_month_totals = (
    monthly_totals_seasonal
    .groupby("Month")["Payment_Amount"]
    .median()
    .reindex(range(1, 13))
)

typical_month_millions = typical_month_totals / 1_000_000

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

print("MEDIAN MONTHLY PAYMENT TOTAL BY CALENDAR MONTH (MILLIONS)")

display(
    typical_month_millions
    .rename(index=dict(enumerate(month_names, start=1)))
    .to_frame(name="Median Monthly Total (Millions)")
    .round(2)
)

plt.figure(figsize=(11, 5))

plt.plot(
    month_names,
    typical_month_millions.values,
    marker="o",
    linewidth=2
)

plt.title("Typical Monthly Payment Totals (2017–2025)")
plt.xlabel("Calendar Month")
plt.ylabel("Median Payment Total (Millions)")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

March recorded the highest median monthly payment total at 24.56 million, while August had the lowest at 13.99 million. December also showed a relatively high median of 22.68 million. These differences indicate variation in monthly spending patterns, although further analysis is needed to establish whether consistent seasonal effects exist.

#### Monthly Transaction Counts by Payment Method

```python
method_count_data = df[
    ["Payment_Date", "payment_method"]
].dropna(subset=["Payment_Date"]).copy()

method_count_data["payment_method"] = (
    method_count_data["payment_method"].fillna("Missing")
)

method_count_data["Year"] = (
    method_count_data["Payment_Date"].dt.year
)

method_count_data["Month"] = (
    method_count_data["Payment_Date"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

method_count_data = method_count_data[
    (method_count_data["Year"] >= 2017) &
    (method_count_data["Year"] <= 2025)
].copy()

# Count transactions for each method by month
monthly_method_counts = (
    method_count_data
    .groupby(["Month", "payment_method"])
    .size()
    .unstack(fill_value=0)
    .sort_index()
)

print("MONTHLY TRANSACTION COUNTS BY PAYMENT METHOD")
display(monthly_method_counts.head())

# Separate charts with a shared time axis
fig, axes = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

# System checks
if "system check" in monthly_method_counts.columns:
    axes[0].plot(
        monthly_method_counts.index,
        monthly_method_counts["system check"],
        linewidth=1.5
    )

axes[0].set_title("Monthly Transaction Counts — System Checks")
axes[0].set_ylabel("Transaction Count")
axes[0].grid(axis="y", linestyle="--", alpha=0.4)

# E-payables
if "epayables" in monthly_method_counts.columns:
    axes[1].plot(
        monthly_method_counts.index,
        monthly_method_counts["epayables"],
        linewidth=1.5
    )

axes[1].set_title("Monthly Transaction Counts — E-Payables")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Transaction Count")
axes[1].grid(axis="y", linestyle="--", alpha=0.4)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

System checks generally recorded approximately 2,000–4,000 transactions per month, with an exceptional spike in December 2018. Validation showed that 12,277 transactions occurred on 21 December 2018. E-payables transaction counts fluctuated considerably and were substantially lower in 2025 than in the earlier period. These patterns highlight differences in transaction volume between the two payment methods.

####  Negative Payment Analysis by Entry Type

```python
negative_payment_data = df.loc[
    df["Payment_Amount"] < 0,
    ["entry_type", "Payment_Amount"]
].copy()

negative_payment_data["entry_type"] = (
    negative_payment_data["entry_type"].fillna("Missing")
)

# Negative payments by entry type
negative_summary = (
    negative_payment_data
    .groupby("entry_type")
    .agg(
        Transaction_Count=("Payment_Amount", "count"),
        Total_Negative_Amount=("Payment_Amount", "sum")
    )
)

# Calculate the share of all negative transactions
negative_summary["Transaction_Share_Percent"] = (
    negative_summary["Transaction_Count"]
    / negative_summary["Transaction_Count"].sum()
    * 100
)

negative_summary["Negative_Amount_Millions"] = (
    -negative_summary["Total_Negative_Amount"] / 1_000_000
)

print("NEGATIVE PAYMENTS BY ENTRY TYPE")

display(
    negative_summary[
        [
            "Transaction_Count",
            "Transaction_Share_Percent",
            "Total_Negative_Amount",
            "Negative_Amount_Millions"
        ]
    ].round(2)
)
```

AP/ accounts for 68.34% of negative transactions (2,372 records), with a combined value of −9.27 million. AP/AD accounts for the remaining 31.66% (1,099 records), totalling −5.95 million. Thus, AP/ has both more negative transactions and a larger total negative value. These records require investigation to determine their underlying purpose.

#### Negative Payment Analysis by Payment Method

```python
negative_method_data = df.loc[
    df["Payment_Amount"] < 0,
    ["payment_method", "Payment_Amount"]
].copy()

negative_method_data["payment_method"] = (
    negative_method_data["payment_method"].fillna("Missing")
)

# Summarize negative transactions by payment method
negative_method_summary = (
    negative_method_data
    .groupby("payment_method")
    .agg(
        Transaction_Count=("Payment_Amount", "count"),
        Total_Negative_Amount=("Payment_Amount", "sum")
    )
)

# Calculate each method's share of negative transactions
negative_method_summary["Transaction_Share_Percent"] = (
    negative_method_summary["Transaction_Count"]
    / negative_method_summary["Transaction_Count"].sum()
    * 100
)

negative_method_summary["Negative_Amount_Millions"] = (
    -negative_method_summary["Total_Negative_Amount"] / 1_000_000
)

print("NEGATIVE PAYMENTS BY PAYMENT METHOD")

display(
    negative_method_summary[
        [
            "Transaction_Count",
            "Transaction_Share_Percent",
            "Total_Negative_Amount",
            "Negative_Amount_Millions"
        ]
    ].round(2)
)
```

System checks account for 83.81% of negative transactions, with a total negative value of 14.83 million, while e-payables account for 16.19%, totalling 0.39 million. Since system checks also dominate overall transaction volume, these figures alone do not establish which method has a higher rate of negative transactions.

#### Negative Transaction Rate by Payment Method

```python
negative_rate_data = df[
    ["payment_method", "Payment_Amount"]
].copy()

negative_rate_data["payment_method"] = (
    negative_rate_data["payment_method"].fillna("Missing")
)

negative_rate_summary = (
    negative_rate_data
    .groupby("payment_method")
    .agg(
        Total_Transactions=("Payment_Amount", "count"),
        Negative_Transactions=(
            "Payment_Amount",
            lambda values: (values < 0).sum()
        ),
        Total_Negative_Amount=(
            "Payment_Amount",
            lambda values: values[values < 0].sum()
        )
    )
)

# The percentage of transactions that are negative
negative_rate_summary["Negative_Transaction_Rate (%)"] = (
    negative_rate_summary["Negative_Transactions"]
    / negative_rate_summary["Total_Transactions"]
    * 100
)

print("NEGATIVE TRANSACTION RATE BY PAYMENT METHOD")

display(negative_rate_summary.round(2))
```

E-payables have a higher negative transaction rate (1.32%) than system checks (0.89%). However, system checks account for more negative transactions (2,909) and a much larger total negative value (14.83 million) because they process substantially more transactions overall. These findings identify differences in negative transaction patterns but do not establish their underlying causes.
